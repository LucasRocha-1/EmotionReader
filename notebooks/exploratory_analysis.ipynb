# EmotionReader – Análise Exploratória

Notebook de exploração e testes rápidos do pipeline de leitura de emoções.

```python
import sys
sys.path.append("..")

from src.pipeline.emotion_pipeline import EmotionPipeline
from src.utils.metrics import emotion_distribution
from src.visualization.plots import plot_emotion_distribution, plot_confidence_bars
```

```python
pipeline = EmotionPipeline(config_path="../config/settings.yaml")
```

## Análise de textos

```python
texts = [
    "Estou extremamente feliz com a aprovação!",
    "Que raiva dessa situação toda.",
    "Estou com medo do que pode acontecer.",
    "Nossa, que surpresa receber essa mensagem.",
    "Me sinto um pouco triste hoje.",
    "O dia está normal, sem grandes novidades."
]

results = pipeline.analyze_text_batch(texts)
for r in results:
    print(f"{r['dominant_emotion']:12} ({r['confidence']:.2f}) ← {r['original_text'][:50]}")
```

```python
dist = emotion_distribution(results)
plot_emotion_distribution(dist, title="Distribuição de Emoções – Exemplos")
```

## Visualizar scores de um exemplo

```python
example = results[0]
plot_confidence_bars(example["all_scores"], title=f"Scores – {example['dominant_emotion']}")
```
